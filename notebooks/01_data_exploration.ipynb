# 01 - Data Exploration

Exploratory data analysis of synthetic protein dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

%matplotlib inline
%config InlineBackend.figure_format = 'retina'

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## Load Data

```python
DATA_PATH = Path('../data/synthetic/example_proteins.csv')
df = pd.read_csv(DATA_PATH)

print(f"Dataset shape: {df.shape}")
print(f"\nColumn names:\n{df.columns.tolist()}")
df.head()
```

## Basic Statistics

```python
print("Dataset Info:")
print(df.info())
print("\nBasic Statistics:")
df.describe()
```

## Check Missing Values

```python
missing = df.isnull().sum()
if missing.sum() > 0:
    print("Missing values found:")
    print(missing[missing > 0])
else:
    print("No missing values!")
```

## Molecular Weight Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(df['molecular_weight'], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Molecular Weight (Da)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Molecular Weights')
axes[0].axvline(df['molecular_weight'].mean(), color='red', linestyle='--', label=f'Mean: {df["molecular_weight"].mean():.2f}')
axes[0].legend()

# Box plot
axes[1].boxplot(df['molecular_weight'])
axes[1].set_ylabel('Molecular Weight (Da)')
axes[1].set_title('Molecular Weight Box Plot')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Expression Levels

```python
expression_counts = df['expression_level'].value_counts()
print("Expression Level Distribution:")
print(expression_counts)

plt.figure(figsize=(8, 6))
expression_counts.plot(kind='bar', color='steelblue', edgecolor='black')
plt.xlabel('Expression Level')
plt.ylabel('Count')
plt.title('Distribution of Protein Expression Levels')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Subcellular Locations

```python
# Parse locations (may have multiple separated by semicolon)
locations = df['subcellular_location'].str.split(';').explode()
location_counts = locations.value_counts().head(10)

plt.figure(figsize=(10, 6))
location_counts.plot(kind='barh', color='coral', edgecolor='black')
plt.xlabel('Count')
plt.ylabel('Subcellular Location')
plt.title('Top 10 Subcellular Locations')
plt.tight_layout()
plt.show()
```

## Disease Associations

```python
disease_counts = df['disease_association'].value_counts().head(10)

plt.figure(figsize=(10, 6))
disease_counts.plot(kind='barh', color='lightcoral', edgecolor='black')
plt.xlabel('Count')
plt.ylabel('Disease Association')
plt.title('Top 10 Disease Associations')
plt.tight_layout()
plt.show()
```

## Summary Statistics Export

```python
# Create summary
summary = {
    'total_proteins': len(df),
    'mean_molecular_weight': df['molecular_weight'].mean(),
    'std_molecular_weight': df['molecular_weight'].std(),
    'unique_diseases': df['disease_association'].nunique(),
    'unique_locations': df['subcellular_location'].nunique()
}

print("\nSummary Statistics:")
for key, value in summary.items():
    print(f"{key}: {value:.2f}" if isinstance(value, float) else f"{key}: {value}")

# Save summary
output_path = Path('../results/tables/01_data_summary.csv')
output_path.parent.mkdir(parents=True, exist_ok=True)
pd.DataFrame([summary]).to_csv(output_path, index=False)
print(f"\nSummary saved to {output_path}")
```
