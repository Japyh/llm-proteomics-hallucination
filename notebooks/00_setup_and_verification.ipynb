## 2. Check Library Versions

```python
libraries = {
    'NumPy': np.__version__,
    'Pandas': pd.__version__,
    'SciPy': scipy.__version__,
    'Matplotlib': matplotlib.__version__,
    'Seaborn': sns.__version__,
    'Plotly': plotly.__version__,
    'Scikit-learn': sklearn.__version__,
    'PyTorch': torch.__version__,
    'Statsmodels': statsmodels.__version__,
    'Biopython': Bio.__version__,
}

version_df = pd.DataFrame(list(libraries.items()), columns=['Library', 'Version'])
print("\nInstalled Library Versions:")
print(version_df.to_string(index=False))
```

## 3. Check GPU Availability

```python
if torch.cuda.is_available():
    print(f"GPU is available!")
    print(f"GPU Device: {torch.cuda.get_device_name(0)}")
    print(f"Number of GPUs: {torch.cuda.device_count()}")
    print(f"CUDA Version: {torch.version.cuda}")
else:
    print("GPU not available. Using CPU.")
    print("This is fine for most analyses, but LLM inference will be slower.")
```

## 4. Verify Data Access

```python
import os
from pathlib import Path

# Define paths
PROJECT_ROOT = Path('..').resolve()
DATA_DIR = PROJECT_ROOT / 'data'
SYNTHETIC_DATA = DATA_DIR / 'synthetic' / 'example_proteins.csv'

print(f"Project root: {PROJECT_ROOT}")
print(f"Data directory exists: {DATA_DIR.exists()}")
print(f"Synthetic data file exists: {SYNTHETIC_DATA.exists()}")

if SYNTHETIC_DATA.exists():
    df = pd.read_csv(SYNTHETIC_DATA)
    print(f"\nSynthetic protein dataset loaded successfully!")
    print(f"Shape: {df.shape}")
    print(f"Columns: {list(df.columns)}")
else:
    print("\nERROR: Synthetic data file not found!")
```

## 5. Test Plotting

```python
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

# Simple test plot
fig, ax = plt.subplots(figsize=(8, 4))
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label='sin(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Test Plot - Environment Verification')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Plotting works correctly!")
```

## 6. Check Environment Variables

```python
from dotenv import load_dotenv

# Load .env file if it exists
env_path = PROJECT_ROOT / '.env'
if env_path.exists():
    load_dotenv(env_path)
    print(".env file loaded successfully!")
    
    # Check for API keys (without printing them)
    api_keys_present = {
        'OPENAI_API_KEY': os.getenv('OPENAI_API_KEY') is not None,
        'ANTHROPIC_API_KEY': os.getenv('ANTHROPIC_API_KEY') is not None,
        'GOOGLE_API_KEY': os.getenv('GOOGLE_API_KEY') is not None,
    }
    
    print("\nAPI Keys Status:")
    for key, present in api_keys_present.items():
        status = "Present" if present else "Missing"
        print(f"  {key}: {status}")
else:
    print(".env file not found.")
    print("API keys will be needed for notebooks 02 and 03.")
    print("Create .env file by copying .env.example")
```
