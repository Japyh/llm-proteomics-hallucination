Create publication-ready figures for The Lancet Digital Health manuscript.

**Objective:** Generate high-quality figures (300 DPI, RGB) summarizing key findings.

**Study Information:**

```python
import sys
sys.path.append('../src')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

np.random.seed(42)
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 300
plt.rcParams['savefig.dpi'] = 300
```

## Generate Figures

```python
output_dir = Path('../paper/figures')
output_dir.mkdir(parents=True, exist_ok=True)

# Figure 1: Model comparison
fig, ax = plt.subplots(1, 1, figsize=(10, 6))
models = ['GPT-4', 'Claude-3', 'Gemini']
rates = [0.15, 0.13, 0.18]
ax.bar(models, rates, color=['#4472C4', '#70AD47', '#FFC000'], edgecolor='black')
ax.set_ylabel('Hallucination Rate')
ax.set_title('Figure 1: Hallucination Rates by Model')
fig.savefig(output_dir / 'Figure_1.png', dpi=300, bbox_inches='tight')
print("Figure 1 saved")
plt.show()
```
